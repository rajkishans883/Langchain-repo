```python
from langchain_classic.retrievers.contextual_compression import ContextualCompressionRetriever
from langchain_classic.retrievers.document_compressors import LLMChainExtractor
from langchain_community.vectorstores import FAISS
from langchain_mistralai import ChatMistralAI, MistralAIEmbeddings
from google.colab import userdata
import os

# Ensure MISTRAL_API_KEY is available
try:
    mistral_api_key = userdata.get('MISTRAL_API_KEY')
    os.environ["MISTRAL_API_KEY"] = mistral_api_key # Also set it in environment for general access
except userdata.SecretNotFoundError:
    print("Error: MISTRAL_API_KEY not found in Secrets. Please add it to the Secrets tab (key icon) and enable notebook access.")
    mistral_api_key = None # Set to None to prevent further errors if key is missing

if mistral_api_key:
    # 1. Initialize your embeddings and Vector Store
    embeddings = MistralAIEmbeddings(mistral_api_key=mistral_api_key)
    vectorstore = FAISS.from_documents(documents, embeddings)

    # 2. Define your retriever variable with a completely unique name
    my_vector_retriever = vectorstore.as_retriever(search_kwargs={"k": 2})

    # 3. Create your LLM compressor engine
    llm = ChatMistralAI(model_name="mistral-tiny", temperature=0, mistral_api_key=mistral_api_key)
    compressor = LLMChainExtractor.from_llm(llm)

    # 4. Pass your unique variable into LangChain's required 'base_retriever' parameter
    compressor_retriever = ContextualCompressionRetriever(
        base_compressor=compressor,
        base_retriever=my_vector_retriever  # LangChain Parameter = Your Variable
    )

    # 5. Invoke it! It will fetch raw docs, compress them, and return only matching context.
    query = "hat is the exact timeout configuration for the vector database?"
    query2="How much magnesium should someone take for better sleep?"
    compressed_docs = compressor_retriever.invoke(query2)

    for doc in compressed_docs:
        print(doc.page_content)
        print("-" * 30)
else:
    print("MistralAI operations skipped due to missing API key.")
```
